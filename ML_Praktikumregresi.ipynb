```python

# ==============================================================
# PENELITIAN REGRESI DATASET ADVERTISING
# Linear Regression, Decision Tree, Random Forest
# ==============================================================

# ==============================================================
# 1. IMPORT LIBRARY
# ==============================================================

import os
import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from google.colab import drive

from sklearn.model_selection import train_test_split, KFold, cross_val_score
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

# Pengaturan visualisasi
sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (10, 6)


# ==============================================================
# 2. MOUNT GOOGLE DRIVE DAN MEMBACA DATASET
# ==============================================================

drive.mount('/content/drive')

file_path = '/content/drive/MyDrive/Advertising.csv'

df = pd.read_csv(file_path)

# Normalisasi nama kolom agar tidak terjadi KeyError
df.columns = df.columns.str.strip().str.lower()

# Menyeragamkan nama kolom
df = df.rename(columns={
    'tv': 'TV',
    'radio': 'Radio',
    'newspaper': 'Newspaper',
    'sales': 'Sales'
})

print("=" * 65)
print("DATASET ADVERTISING")
print("=" * 65)

print("\nNama Kolom:", df.columns.tolist())
print("Ukuran Dataset:", df.shape)

print("\nInformasi Dataset:")
df.info()

print("\nLima Data Pertama:")
display(df.head())

print("\nStatistik Deskriptif:")
display(df.describe())


# ==============================================================
# 3. DATA PREPROCESSING
# ==============================================================

# Menghapus kolom indeks yang tidak diperlukan
df = df.loc[:, ~df.columns.str.contains('^unnamed', case=False)]

# Memastikan seluruh kolom penelitian berbentuk numerik
features = ['TV', 'Radio', 'Newspaper']
target = 'Sales'

for col in features + [target]:
    df[col] = pd.to_numeric(df[col], errors='coerce')

# Memeriksa missing value
print("\nMissing Value Sebelum Preprocessing:")
print(df[features + [target]].isnull().sum())

# Menghapus baris yang memiliki missing value
df = df.dropna(subset=features + [target])

# Menghapus data duplikat
df = df.drop_duplicates()

# Menentukan variabel independen dan dependen
X = df[features]
y = df[target]

print("\nJumlah Data Setelah Preprocessing:", len(df))
print("Variabel Independen:", features)
print("Variabel Dependen:", target)

print("\nData Setelah Preprocessing:")
display(df.head())


# ==============================================================
# 4. ANALISIS KORELASI PEARSON
# ==============================================================

correlation = df[features + [target]].corr(method='pearson')

print("\nMatriks Korelasi Pearson:")
display(correlation.round(4))

plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    fmt='.3f',
    linewidths=0.5,
    square=True
)

plt.title('Heatmap Korelasi Pearson Advertising')
plt.tight_layout()
plt.show()


# ==============================================================
# 5. ANALISIS DISTRIBUSI DATASET
# ==============================================================

fig, axes = plt.subplots(1, 4, figsize=(18, 4))

for i, col in enumerate(features + [target]):

    sns.histplot(
        df[col],
        kde=True,
        ax=axes[i]
    )

    axes[i].set_title(f'Distribusi {col}')
    axes[i].set_xlabel(col)
    axes[i].set_ylabel('Frekuensi')

plt.suptitle('Distribusi Variabel Dataset Advertising')
plt.tight_layout()
plt.show()


# ==============================================================
# 6. PEMBAGIAN DATA TRAINING DAN TESTING
# ==============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("\n" + "=" * 65)
print("PEMBAGIAN DATASET")
print("=" * 65)

print("Total Dataset :", len(df))
print("Data Training:", len(X_train))
print("Data Testing :", len(X_test))


# ==============================================================
# 7. MEMBANGUN MODEL REGRESI
# ==============================================================

models = {

    'Linear Regression': LinearRegression(),

    'Decision Tree Regression': DecisionTreeRegressor(
        random_state=42
    ),

    'Random Forest Regression': RandomForestRegressor(
        n_estimators=100,
        random_state=42,
        n_jobs=-1
    )

}


# ==============================================================
# 8. TRAINING DAN EVALUASI MODEL
# ==============================================================

results = []
predictions = {}
trained_models = {}

# Cross Validation 5-Fold
kfold = KFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

for name, model in models.items():

    print(f"\nProses Training: {name}")

    # Training model
    model.fit(X_train, y_train)

    trained_models[name] = model

    # Prediksi data testing
    y_pred = model.predict(X_test)

    predictions[name] = y_pred

    # Perhitungan metrik evaluasi
    mae = mean_absolute_error(y_test, y_pred)

    mse = mean_squared_error(y_test, y_pred)

    rmse = np.sqrt(mse)

    r2 = r2_score(y_test, y_pred)

    # Cross Validation
    cv_scores = cross_val_score(
        model,
        X,
        y,
        cv=kfold,
        scoring='r2',
        n_jobs=-1
    )

    # Menyimpan hasil evaluasi
    results.append({

        'Model': name,
        'MAE': mae,
        'MSE': mse,
        'RMSE': rmse,
        'R2 Score': r2,
        'CV R2 Mean': cv_scores.mean(),
        'CV R2 Std': cv_scores.std()

    })

# Membuat tabel evaluasi
results_df = pd.DataFrame(results)

print("\n" + "=" * 80)
print("HASIL EVALUASI MODEL REGRESI")
print("=" * 80)

display(results_df.round(5))


# ==============================================================
# 9. MENENTUKAN MODEL TERBAIK
# ==============================================================

best_index = results_df['R2 Score'].idxmax()

best_model_name = results_df.loc[best_index, 'Model']

best_row = results_df.loc[best_index]

print("\nMODEL DENGAN R2 SCORE TERTINGGI")
print("Model:", best_model_name)
print("R2 Score:", round(best_row['R2 Score'], 5))
print("MAE:", round(best_row['MAE'], 5))
print("RMSE:", round(best_row['RMSE'], 5))


# ==============================================================
# 10. VISUALISASI PERBANDINGAN METRIK EVALUASI
# ==============================================================

metrics = ['MAE', 'MSE', 'RMSE', 'R2 Score']

fig, axes = plt.subplots(2, 2, figsize=(14, 10))

axes = axes.flatten()

for i, metric in enumerate(metrics):

    sns.barplot(
        data=results_df,
        x='Model',
        y=metric,
        ax=axes[i]
    )

    axes[i].set_title(f'Perbandingan {metric}')
    axes[i].set_xlabel('Algoritma')
    axes[i].set_ylabel(metric)
    axes[i].tick_params(axis='x', rotation=15)

plt.suptitle(
    'Perbandingan Kinerja Algoritma Regresi',
    fontsize=15
)

plt.tight_layout()
plt.show()


# ==============================================================
# 11. VISUALISASI ACTUAL VS PREDICTED
# ==============================================================

fig, axes = plt.subplots(1, 3, figsize=(18, 5))

for i, (name, y_pred) in enumerate(predictions.items()):

    axes[i].scatter(
        y_test,
        y_pred,
        alpha=0.7
    )

    min_val = min(y_test.min(), y_pred.min())
    max_val = max(y_test.max(), y_pred.max())

    axes[i].plot(
        [min_val, max_val],
        [min_val, max_val],
        'r--'
    )

    axes[i].set_title(name)
    axes[i].set_xlabel('Actual Sales')
    axes[i].set_ylabel('Predicted Sales')

plt.suptitle('Actual vs Predicted Sales')
plt.tight_layout()
plt.show()


# ==============================================================
# 12. ANALISIS RESIDUAL
# ==============================================================

fig, axes = plt.subplots(1, 3, figsize=(18, 5))

for i, (name, y_pred) in enumerate(predictions.items()):

    residuals = y_test.values - y_pred

    sns.scatterplot(
        x=y_pred,
        y=residuals,
        ax=axes[i]
    )

    axes[i].axhline(
        y=0,
        color='red',
        linestyle='--'
    )

    axes[i].set_title(name)
    axes[i].set_xlabel('Predicted Sales')
    axes[i].set_ylabel('Residual')

plt.suptitle('Residual Analysis')
plt.tight_layout()
plt.show()


# ==============================================================
# 13. DISTRIBUSI RESIDUAL
# ==============================================================

fig, axes = plt.subplots(1, 3, figsize=(18, 5))

for i, (name, y_pred) in enumerate(predictions.items()):

    residuals = y_test.values - y_pred

    sns.histplot(
        residuals,
        kde=True,
        ax=axes[i]
    )

    axes[i].set_title(name)
    axes[i].set_xlabel('Residual')
    axes[i].set_ylabel('Frekuensi')

plt.suptitle('Residual Distribution')
plt.tight_layout()
plt.show()


# ==============================================================
# 14. FEATURE IMPORTANCE DECISION TREE DAN RANDOM FOREST
# ==============================================================

dt_model = trained_models['Decision Tree Regression']

rf_model = trained_models['Random Forest Regression']

dt_importance = pd.DataFrame({

    'Feature': features,
    'Importance': dt_model.feature_importances_

})

rf_importance = pd.DataFrame({

    'Feature': features,
    'Importance': rf_model.feature_importances_

})

fig, axes = plt.subplots(1, 2, figsize=(13, 5))

sns.barplot(
    data=dt_importance,
    x='Importance',
    y='Feature',
    ax=axes[0]
)

axes[0].set_title('Decision Tree Feature Importance')

sns.barplot(
    data=rf_importance,
    x='Importance',
    y='Feature',
    ax=axes[1]
)

axes[1].set_title('Random Forest Feature Importance')

plt.tight_layout()
plt.show()

print("\nFeature Importance Decision Tree:")
display(
    dt_importance.sort_values(
        by='Importance',
        ascending=False
    )
)

print("\nFeature Importance Random Forest:")
display(
    rf_importance.sort_values(
        by='Importance',
        ascending=False
    )
)


# ==============================================================
# 15. KOEFISIEN LINEAR REGRESSION
# ==============================================================

lr_model = trained_models['Linear Regression']

coefficients = pd.DataFrame({

    'Feature': features,
    'Coefficient': lr_model.coef_

})

print("\nKoefisien Linear Regression:")
display(coefficients)

print("Intercept:", lr_model.intercept_)


# ==============================================================
# 16. TABEL PERBANDINGAN AKTUAL DAN PREDIKSI
# ==============================================================

prediction_df = pd.DataFrame({

    'Actual Sales': y_test.values

})

for name, y_pred in predictions.items():

    prediction_df[name] = y_pred

print("\nHasil Aktual dan Prediksi:")
display(prediction_df.head(20))


# ==============================================================
# 17. RINGKASAN HASIL PENELITIAN
# ==============================================================

print("\n" + "=" * 70)
print("RINGKASAN HASIL PENELITIAN")
print("=" * 70)

for _, row in results_df.iterrows():

    print(f"\nModel: {row['Model']}")

    print(f"MAE      : {row['MAE']:.5f}")
    print(f"MSE      : {row['MSE']:.5f}")
    print(f"RMSE     : {row['RMSE']:.5f}")
    print(f"R2 Score : {row['R2 Score']:.5f}")

    print(f"CV Mean  : {row['CV R2 Mean']:.5f}")
    print(f"CV Std   : {row['CV R2 Std']:.5f}")

print("\nModel dengan R2 Score tertinggi:", best_model_name)


# ==============================================================
# 18. MENYIMPAN HASIL PENELITIAN KE GOOGLE DRIVE
# ==============================================================

output_dir = '/content/drive/MyDrive/Hasil_Regresi_Advertising'

os.makedirs(output_dir, exist_ok=True)

# Simpan hasil evaluasi
results_df.to_csv(
    f'{output_dir}/hasil_evaluasi_model.csv',
    index=False
)

# Simpan hasil prediksi
prediction_df.to_csv(
    f'{output_dir}/hasil_prediksi.csv',
    index=False
)

# Simpan matriks korelasi
correlation.to_csv(
    f'{output_dir}/korelasi_pearson.csv'
)

# Simpan feature importance
dt_importance.to_csv(
    f'{output_dir}/feature_importance_decision_tree.csv',
    index=False
)

rf_importance.to_csv(
    f'{output_dir}/feature_importance_random_forest.csv',
    index=False
)

# Simpan koefisien Linear Regression
coefficients.to_csv(
    f'{output_dir}/koefisien_linear_regression.csv',
    index=False
)

# Simpan seluruh hasil prediksi
print("\n" + "=" * 70)
print("EKSPERIMEN SELESAI")
print("=" * 70)

print("Seluruh hasil penelitian berhasil disimpan.")
print("Lokasi:", output_dir)

print("\nDaftar File:")

for filename in os.listdir(output_dir):
    print("-", filename)
```
