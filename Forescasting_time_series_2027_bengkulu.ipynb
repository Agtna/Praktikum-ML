```python

# ==============================================================
# TIME SERIES FORECASTING LUAS KEBAKARAN HUTAN DAN LAHAN
# PROVINSI BENGKULU MENGGUNAKAN ARIMA
# DATA HISTORIS: 2000-2026
# PERIODE FORECASTING: 2027
# ==============================================================


# ==============================================================
# 1. INSTALL DAN IMPORT LIBRARY
# ==============================================================

!pip install -q statsmodels openpyxl scikit-learn

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import warnings

from google.colab import drive, files
from statsmodels.tsa.arima.model import ARIMA
from sklearn.metrics import mean_absolute_error, mean_squared_error
from IPython.display import display

warnings.filterwarnings("ignore")

plt.rcParams['figure.figsize'] = (12, 5)
plt.rcParams['axes.grid'] = True
plt.rcParams['grid.alpha'] = 0.3

print("=" * 70)
print("TIME SERIES FORECASTING KARHUTLA PROVINSI BENGKULU")
print("MENGGUNAKAN ALGORITMA ARIMA")
print("=" * 70)


# ==============================================================
# 2. MEMBACA DATASET
# ==============================================================

drive.mount('/content/drive')

file_path = '/content/drive/MyDrive/luas-karhutla-provinsi (1).xlsx'

df = pd.read_excel(file_path)

print("\nINFORMASI DATASET")
print("-" * 70)

print("Ukuran dataset:", df.shape)
print("Kolom dataset:", df.columns.tolist())

display(df.head())


# ==============================================================
# 3. PREPROCESSING DATA
# ==============================================================

# Memilih data Provinsi Bengkulu

if 'Provinsi' not in df.columns:
    raise ValueError("Kolom Provinsi tidak ditemukan dalam dataset.")

data_bengkulu = df[
    df['Provinsi'].astype(str).str.strip().str.lower() == 'bengkulu'
].copy()

if len(data_bengkulu) != 1:
    raise ValueError(
        "Data Bengkulu tidak ditemukan atau terdapat lebih dari satu baris."
    )

# Mengidentifikasi kolom tahun

year_columns = [
    col for col in df.columns
    if str(col).strip().isdigit()
    and 1900 <= int(str(col).strip()) <= 2100
]

year_columns = sorted(
    year_columns,
    key=lambda x: int(str(x).strip())
)

years = [int(str(col).strip()) for col in year_columns]

if len(years) == 0:
    raise ValueError("Kolom tahun tidak ditemukan.")

# Mengonversi angka dengan format Indonesia
# Contoh: 2.871,00 menjadi 2871.00

def convert_number(value):

    if pd.isna(value):
        return np.nan

    if isinstance(value, (int, float, np.number)):
        return float(value)

    value = str(value).strip()

    if value == '' or value.lower() in ['nan', 'none', '-']:
        return np.nan

    value = value.replace('.', '').replace(',', '.')

    return float(value)


# Mengambil data luas kebakaran

values = [
    convert_number(data_bengkulu.iloc[0][col])
    for col in year_columns
]

# Membentuk data time series menggunakan indeks tahun berurutan

ts = pd.Series(
    values,
    index=pd.RangeIndex(
        start=min(years),
        stop=max(years) + 1,
        name='Tahun'
    ),
    name='Luas_Kebakaran',
    dtype=float
)

# Validasi dataset

if ts.isna().any():
    print("Missing value ditemukan:")
    print(ts[ts.isna()])
    raise ValueError(
        "Terdapat missing value. Periksa data sebelum forecasting."
    )

if (ts < 0).any():
    raise ValueError("Ditemukan nilai luas kebakaran negatif.")

if len(ts) != max(years) - min(years) + 1:
    raise ValueError("Terdapat tahun yang hilang dalam dataset.")

print("\nHASIL PREPROCESSING")
print("-" * 70)

print("Wilayah:", "Provinsi Bengkulu")
print("Periode:", f"{ts.index.min()}-{ts.index.max()}")
print("Jumlah observasi:", len(ts))
print("Missing value:", ts.isna().sum())

display(ts.to_frame())


# ==============================================================
# 4. TABEL 1 - KARAKTERISTIK DATASET
# ==============================================================

tabel_1 = pd.DataFrame({
    'Karakteristik': [
        'Sumber Data',
        'Wilayah Penelitian',
        'Periode Data Historis',
        'Jumlah Observasi',
        'Variabel Penelitian',
        'Jenis Data',
        'Frekuensi Data',
        'Satuan',
        'Tahun Forecasting'
    ],

    'Keterangan': [
        'SIPONGI',
        'Provinsi Bengkulu',
        f'{ts.index.min()}-{ts.index.max()}',
        len(ts),
        'Luas Kebakaran Hutan dan Lahan',
        'Univariate Time Series',
        'Tahunan',
        'Hektare (ha)',
        2027
    ]
})

print("\n" + "=" * 70)
print("TABEL 1. KARAKTERISTIK DATASET")
print("=" * 70)

display(tabel_1)


# ==============================================================
# 5. TABEL 2 - STATISTIK DESKRIPTIF
# ==============================================================

tabel_2 = pd.DataFrame({
    'Statistik': [
        'Jumlah Observasi',
        'Nilai Minimum',
        'Nilai Maksimum',
        'Rata-rata',
        'Median',
        'Standar Deviasi'
    ],

    'Nilai': [
        ts.count(),
        ts.min(),
        ts.max(),
        ts.mean(),
        ts.median(),
        ts.std()
    ]
})

tabel_2['Nilai'] = tabel_2['Nilai'].round(2)

print("\n" + "=" * 70)
print("TABEL 2. STATISTIK DESKRIPTIF")
print("=" * 70)

display(tabel_2)


# ==============================================================
# 6. GRAFIK 1 - TREN DATA HISTORIS
# ==============================================================

plt.figure(figsize=(13, 5))

plt.plot(
    ts.index,
    ts.values,
    marker='o',
    linewidth=2,
    markersize=5,
    label='Luas Kebakaran'
)

plt.title(
    'Tren Luas Kebakaran Hutan dan Lahan Provinsi Bengkulu (2000-2026)',
    fontweight='bold'
)

plt.xlabel('Tahun')
plt.ylabel('Luas Kebakaran (Hektare)')
plt.xticks(ts.index, rotation=45)
plt.legend()

plt.tight_layout()
plt.show()


# ==============================================================
# 7. PEMBAGIAN DATA TRAINING DAN TESTING
# ==============================================================

# Pembagian data secara kronologis 80:20

split_index = int(len(ts) * 0.80)

train = ts.iloc[:split_index]
test = ts.iloc[split_index:]

# Transformasi log1p untuk mengurangi pengaruh nilai ekstrem
# Transformasi ini juga dapat menangani nilai nol.

train_log = np.log1p(train)

print("\nINFORMASI PEMBAGIAN DATA")
print("-" * 70)

print("Total observasi:", len(ts))
print("Training:", len(train))
print("Testing:", len(test))

print(
    "Periode training:",
    f"{train.index.min()}-{train.index.max()}"
)

print(
    "Periode testing:",
    f"{test.index.min()}-{test.index.max()}"
)


# ==============================================================
# 8. TABEL 3 - PEMBAGIAN DATA
# ==============================================================

tabel_3 = pd.DataFrame({
    'Jenis Data': [
        'Training',
        'Testing',
        'Total'
    ],

    'Tahun Awal': [
        train.index.min(),
        test.index.min(),
        ts.index.min()
    ],

    'Tahun Akhir': [
        train.index.max(),
        test.index.max(),
        ts.index.max()
    ],

    'Jumlah Observasi': [
        len(train),
        len(test),
        len(ts)
    ],

    'Persentase (%)': [
        len(train) / len(ts) * 100,
        len(test) / len(ts) * 100,
        100
    ]
})

tabel_3['Persentase (%)'] = tabel_3['Persentase (%)'].round(2)

print("\n" + "=" * 70)
print("TABEL 3. PEMBAGIAN DATA TRAINING DAN TESTING")
print("=" * 70)

display(tabel_3)


# ==============================================================
# 9. GRAFIK 2 - VISUALISASI TRAINING DAN TESTING
# ==============================================================

plt.figure(figsize=(13, 5))

plt.plot(
    train.index,
    train.values,
    marker='o',
    linewidth=2,
    label='Data Training'
)

plt.plot(
    test.index,
    test.values,
    marker='o',
    linewidth=2,
    label='Data Testing'
)

plt.axvline(
    x=test.index.min(),
    linestyle='--',
    label='Batas Training dan Testing'
)

plt.title(
    'Pembagian Data Training dan Testing',
    fontweight='bold'
)

plt.xlabel('Tahun')
plt.ylabel('Luas Kebakaran (Hektare)')
plt.legend()

plt.tight_layout()
plt.show()


# ==============================================================
# 10. PEMILIHAN PARAMETER MODEL ARIMA
# ==============================================================

# Grid search parameter ARIMA
# Pemilihan model berdasarkan nilai AIC terkecil.

p_values = range(0, 3)
d_values = range(0, 2)
q_values = range(0, 3)

hasil_parameter = []

best_aic = np.inf
best_order = None
best_model = None

print("\n" + "=" * 70)
print("PROSES PEMILIHAN PARAMETER ARIMA")
print("=" * 70)

for p in p_values:

    for d in d_values:

        for q in q_values:

            order = (p, d, q)

            try:

                trend = 'c' if d == 0 else 't'

                model = ARIMA(
                    train_log,
                    order=order,
                    trend=trend
                )

                fitted_model = model.fit()

                aic = fitted_model.aic

                if np.isfinite(aic):

                    hasil_parameter.append({
                        'Parameter ARIMA': str(order),
                        'AIC': aic
                    })

                    if aic < best_aic:

                        best_aic = aic
                        best_order = order
                        best_model = fitted_model

            except Exception:
                continue

if best_model is None:
    raise RuntimeError("Tidak ditemukan model ARIMA yang berhasil.")

tabel_parameter = pd.DataFrame(hasil_parameter)

tabel_parameter = tabel_parameter.sort_values(
    by='AIC'
).reset_index(drop=True)

print("\nPARAMETER TERBAIK")
print("-" * 70)

print("ARIMA:", best_order)
print("AIC:", round(best_aic, 4))

display(tabel_parameter.round(4))


# ==============================================================
# 11. FORECASTING PADA DATA TESTING
# ==============================================================

# Melakukan forecasting sepanjang periode testing

forecast_test = best_model.get_forecast(
    steps=len(test)
)

# Mengambil hasil peramalan pada skala log

forecast_test_mean = np.asarray(
    forecast_test.predicted_mean
)

forecast_test_var = np.asarray(
    forecast_test.var_pred_mean
)

# Mengembalikan hasil ke skala hektare
# Menggunakan koreksi bias retransformation.

forecast_test_values = np.expm1(
    forecast_test_mean + 0.5 * forecast_test_var
)

forecast_test_values = np.maximum(
    forecast_test_values,
    0
)

# Membuat tabel hasil forecasting testing

hasil_forecasting_testing = pd.DataFrame({
    'Tahun': test.index,
    'Data Aktual (ha)': test.values,
    'Hasil Forecasting ARIMA (ha)': forecast_test_values
})

hasil_forecasting_testing['Selisih'] = (
    hasil_forecasting_testing['Data Aktual (ha)'] -
    hasil_forecasting_testing['Hasil Forecasting ARIMA (ha)']
)

hasil_forecasting_testing['Absolute Error'] = (
    hasil_forecasting_testing['Selisih'].abs()
)

print("\nHASIL FORECASTING DATA TESTING")
print("-" * 70)

display(hasil_forecasting_testing.round(2))


# ==============================================================
# 12. GRAFIK 3 - ACTUAL VS FORECASTING
# ==============================================================

plt.figure(figsize=(13, 5))

plt.plot(
    test.index,
    test.values,
    marker='o',
    linewidth=2,
    label='Data Aktual'
)

plt.plot(
    test.index,
    forecast_test_values,
    marker='s',
    linestyle='--',
    linewidth=2,
    label='Hasil Forecasting ARIMA'
)

plt.title(
    'Perbandingan Data Aktual dan Hasil Forecasting ARIMA',
    fontweight='bold'
)

plt.xlabel('Tahun')
plt.ylabel('Luas Kebakaran (Hektare)')
plt.xticks(test.index)
plt.legend()

plt.tight_layout()
plt.show()


# ==============================================================
# 13. TABEL 4 - EVALUASI MODEL FORECASTING
# ==============================================================

# MAE

mae = mean_absolute_error(
    test.values,
    forecast_test_values
)

# RMSE

rmse = np.sqrt(
    mean_squared_error(
        test.values,
        forecast_test_values
    )
)

# MASE

naive_error = np.mean(
    np.abs(np.diff(train.values))
)

if naive_error > 0:
    mase = mae / naive_error
else:
    mase = np.nan

tabel_4 = pd.DataFrame({
    'Model': ['ARIMA'],
    'Parameter': [str(best_order)],
    'MAE': [mae],
    'RMSE': [rmse],
    'MASE': [mase]
})

print("\n" + "=" * 70)
print("TABEL 4. EVALUASI MODEL FORECASTING")
print("=" * 70)

display(tabel_4.round(4))


# ==============================================================
# 14. PEMODELAN ULANG MENGGUNAKAN SELURUH DATA
# ==============================================================

# Setelah evaluasi, model dilatih ulang menggunakan
# seluruh data historis tahun 2000-2026.

full_log = np.log1p(ts)

final_trend = 'c' if best_order[1] == 0 else 't'

final_model = ARIMA(
    full_log,
    order=best_order,
    trend=final_trend
).fit()

print("\nMODEL ARIMA FINAL")
print("-" * 70)

print(final_model.summary())


# ==============================================================
# 15. FORECASTING LUAS KARHUTLA TAHUN 2027
# ==============================================================

forecast_2027_log = final_model.get_forecast(
    steps=1
)

# Nilai rata-rata hasil forecasting pada skala log

forecast_mean_log = float(
    forecast_2027_log.predicted_mean.iloc[0]
)

forecast_var_log = float(
    forecast_2027_log.var_pred_mean.iloc[0]
)

# Mengembalikan ke skala hektare dengan koreksi bias

forecast_2027 = np.expm1(
    forecast_mean_log + 0.5 * forecast_var_log
)

forecast_2027 = max(0, forecast_2027)

# Menghitung interval prediksi 95%

forecast_interval_log = forecast_2027_log.conf_int(
    alpha=0.05
)

lower_2027 = np.expm1(
    float(forecast_interval_log.iloc[0, 0])
)

upper_2027 = np.expm1(
    float(forecast_interval_log.iloc[0, 1])
)

lower_2027 = max(0, lower_2027)
upper_2027 = max(0, upper_2027)


# ==============================================================
# 16. TABEL 5 - HASIL FORECASTING TAHUN 2027
# ==============================================================

tabel_5 = pd.DataFrame({
    'Model': ['ARIMA'],
    'Parameter ARIMA': [str(best_order)],
    'Tahun Forecasting': [2027],
    'Hasil Forecasting (ha)': [forecast_2027],
    'Batas Bawah 95% (ha)': [lower_2027],
    'Batas Atas 95% (ha)': [upper_2027]
})

print("\n" + "=" * 70)
print("TABEL 5. HASIL FORECASTING TAHUN 2027")
print("=" * 70)

display(tabel_5.round(2))

print("\nHASIL PERAMALAN")
print("-" * 70)

print(f"Forecasting 2027: {forecast_2027:,.2f} hektare")

print(
    f"Interval prediksi 95%: "
    f"{lower_2027:,.2f} - {upper_2027:,.2f} hektare"
)


# ==============================================================
# 17. GRAFIK 4 - VISUALISASI FORECASTING TAHUN 2027
# ==============================================================

plt.figure(figsize=(13, 6))

# Data historis

plt.plot(
    ts.index,
    ts.values,
    marker='o',
    linewidth=2,
    label='Data Historis 2000-2026'
)

# Garis menuju tahun 2027

plt.plot(
    [2026, 2027],
    [ts.loc[2026], forecast_2027],
    marker='s',
    linestyle='--',
    linewidth=2,
    label='Forecasting Tahun 2027'
)

# Interval forecasting

plt.vlines(
    x=2027,
    ymin=lower_2027,
    ymax=upper_2027,
    linewidth=3,
    label='Interval Prediksi 95%'
)

plt.scatter(
    2027,
    forecast_2027,
    marker='s',
    s=100,
    zorder=5
)

plt.annotate(
    f'{forecast_2027:,.2f} ha',
    xy=(2027, forecast_2027),
    xytext=(10, 12),
    textcoords='offset points'
)

plt.title(
    'Forecasting Luas Kebakaran Hutan dan Lahan Provinsi Bengkulu Tahun 2027',
    fontweight='bold'
)

plt.xlabel('Tahun')
plt.ylabel('Luas Kebakaran (Hektare)')

plt.xticks(
    list(ts.index) + [2027],
    rotation=45
)

plt.legend()
plt.tight_layout()
plt.show()


# ==============================================================
# 18. RINGKASAN HASIL FORECASTING
# ==============================================================

print("\n" + "=" * 70)
print("RINGKASAN HASIL TIME SERIES FORECASTING")
print("=" * 70)

print("Wilayah Penelitian       : Provinsi Bengkulu")
print("Sumber Data              : SIPONGI")
print(f"Periode Data Historis    : {ts.index.min()}-{ts.index.max()}")
print(f"Jumlah Observasi         : {len(ts)}")
print("Metode                   : Time Series Forecasting")
print("Algoritma                : ARIMA")
print(f"Parameter Terbaik        : {best_order}")
print(f"Nilai AIC                : {best_aic:.4f}")
print(f"MAE                      : {mae:.4f}")
print(f"RMSE                     : {rmse:.4f}")
print(f"MASE                     : {mase:.4f}")
print(f"Forecasting Tahun 2027   : {forecast_2027:.2f} ha")
print(
    f"Interval 95%             : "
    f"{lower_2027:.2f} - {upper_2027:.2f} ha"
)

print("=" * 70)


# ==============================================================
# 19. EXPORT HASIL FORECASTING KE EXCEL
# ==============================================================

output_path = '/content/hasil_forecasting_karhutla_bengkulu.xlsx'

with pd.ExcelWriter(
    output_path,
    engine='openpyxl'
) as writer:

    tabel_1.to_excel(
        writer,
        sheet_name='Karakteristik Dataset',
        index=False
    )

    tabel_2.to_excel(
        writer,
        sheet_name='Statistik Deskriptif',
        index=False
    )

    tabel_3.to_excel(
        writer,
        sheet_name='Pembagian Data',
        index=False
    )

    tabel_parameter.to_excel(
        writer,
        sheet_name='Parameter ARIMA',
        index=False
    )

    hasil_forecasting_testing.to_excel(
        writer,
        sheet_name='Forecasting Testing',
        index=False
    )

    tabel_4.to_excel(
        writer,
        sheet_name='Evaluasi Forecasting',
        index=False
    )

    tabel_5.to_excel(
        writer,
        sheet_name='Forecasting 2027',
        index=False
    )

print("\nFILE EXCEL BERHASIL DIBUAT:")
print(output_path)

files.download(output_path)

print("\nPROSES TIME SERIES FORECASTING SELESAI!")
```
