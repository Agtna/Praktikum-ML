```python

# ============================================================
# KLASIFIKASI BANK MARKETING
# ANALISIS PENGARUH SELEKSI FITUR PEARSON
# ALGORITMA KNN, DECISION TREE, DAN RANDOM FOREST
# STRATIFIED 10-FOLD CROSS VALIDATION
# ============================================================


# ============================================================
# 1. IMPORT LIBRARY
# ============================================================

import os
import copy
import warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from scipy import stats

from sklearn.model_selection import StratifiedKFold
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    balanced_accuracy_score,
    confusion_matrix,
    roc_auc_score,
    roc_curve
)

from IPython.display import display

warnings.filterwarnings("ignore")
sns.set_style("whitegrid")


# ============================================================
# 2. KONFIGURASI
# ============================================================

RANDOM_STATE = 42
N_SPLITS = 10
PEARSON_THRESHOLD = 0.05

from google.colab import drive
drive.mount("/content/drive")

DATA_PATH = "/content/drive/MyDrive/bank-full.csv"


# ============================================================
# 3. MEMBACA DATASET
# ============================================================

df = pd.read_csv(DATA_PATH, sep=";")

print("=" * 65)
print("INFORMASI DATASET BANK MARKETING")
print("=" * 65)

print("\nUkuran Dataset:")
print("Jumlah Baris :", df.shape[0])
print("Jumlah Kolom :", df.shape[1])

print("\nLima Data Pertama:")
display(df.head())

print("\nInformasi Dataset:")
df.info()

print("\nJumlah Missing Values:")
display(df.isnull().sum())


# ============================================================
# 4. KARAKTERISTIK DAN DISTRIBUSI DATASET
# ============================================================

target_column = "y"

if target_column not in df.columns:
    raise ValueError("Kolom target 'y' tidak ditemukan.")

X = df.drop(columns=[target_column])
y = df[target_column].astype(str)

dataset_info = pd.DataFrame({
    "Informasi": [
        "Jumlah Data",
        "Jumlah Fitur",
        "Jumlah Kelas Target",
        "Jumlah Data Kosong"
    ],
    "Nilai": [
        len(df),
        X.shape[1],
        y.nunique(),
        df.isnull().sum().sum()
    ]
})

print("\nKarakteristik Dataset:")
display(dataset_info)

# Distribusi kelas
class_distribution = y.value_counts().reset_index()
class_distribution.columns = ["Kelas", "Jumlah"]

class_distribution["Persentase (%)"] = (
    class_distribution["Jumlah"] / len(y) * 100
)

print("\nDistribusi Kelas Target:")
display(class_distribution)

plt.figure(figsize=(7, 5))

sns.countplot(
    x=y,
    order=y.value_counts().index
)

plt.title("Distribusi Kelas Target Bank Marketing")
plt.xlabel("Kelas Target")
plt.ylabel("Jumlah Data")
plt.tight_layout()
plt.show()


# ============================================================
# 5. IDENTIFIKASI TIPE FITUR
# ============================================================

numeric_features = X.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

categorical_features = X.select_dtypes(
    include=["object", "category", "bool"]
).columns.tolist()

feature_types = pd.DataFrame({
    "Jenis Fitur": ["Numerik", "Kategorikal"],
    "Jumlah": [
        len(numeric_features),
        len(categorical_features)
    ]
})

print("\nJenis Fitur:")
display(feature_types)

print("\nFitur Numerik:")
print(numeric_features)

print("\nFitur Kategorikal:")
print(categorical_features)

descriptive_stats = X[numeric_features].describe().T

print("\nStatistik Deskriptif Fitur Numerik:")
display(descriptive_stats)


# ============================================================
# 6. PREPROCESSING DATA
# ============================================================

numeric_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median"))
])

categorical_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, numeric_features),
        ("cat", categorical_transformer, categorical_features)
    ],
    remainder="drop",
    verbose_feature_names_out=False
)


# ============================================================
# 7. DEFINISI ALGORITMA KLASIFIKASI
# ============================================================

models = {
    "KNN": KNeighborsClassifier(
        n_neighbors=5,
        metric="minkowski",
        p=2
    ),

    "Decision Tree": DecisionTreeClassifier(
        criterion="gini",
        max_depth=8,
        random_state=RANDOM_STATE
    ),

    "Random Forest": RandomForestClassifier(
        n_estimators=100,
        max_depth=10,
        random_state=RANDOM_STATE,
        n_jobs=-1
    )
}


# ============================================================
# 8. FUNGSI SELEKSI FITUR PEARSON
# ============================================================

def pearson_feature_selection(
    X_train,
    y_train,
    threshold=0.05
):

    y_numeric = np.asarray(y_train).astype(float)

    correlations = []

    for i in range(X_train.shape[1]):

        feature = X_train[:, i]

        if np.std(feature) == 0 or np.std(y_numeric) == 0:
            corr = 0
        else:
            corr = np.corrcoef(feature, y_numeric)[0, 1]

        if not np.isfinite(corr):
            corr = 0

        correlations.append(corr)

    correlations = np.array(correlations)

    selected_indices = np.where(
        np.abs(correlations) >= threshold
    )[0]

    if len(selected_indices) == 0:
        selected_indices = np.array([
            np.argmax(np.abs(correlations))
        ])

    return selected_indices, correlations


# ============================================================
# 9. STRATIFIED 10-FOLD CROSS VALIDATION
# ============================================================

skf = StratifiedKFold(
    n_splits=N_SPLITS,
    shuffle=True,
    random_state=RANDOM_STATE
)

results = []
feature_records = []
confusion_totals = {}
roc_records = []

classes = np.unique(y)

if len(classes) != 2:
    raise ValueError(
        "Program ini membutuhkan target dengan dua kelas."
    )

positive_label = (
    "yes" if "yes" in classes else classes[-1]
)

y_binary = (y == positive_label).astype(int).to_numpy()

print("\nKelas Positif:", positive_label)


# ============================================================
# 10. PROSES EKSPERIMEN KLASIFIKASI
# ============================================================

for fold, (train_index, test_index) in enumerate(
    skf.split(X, y_binary),
    start=1
):

    print(f"\nMemproses Fold {fold}/{N_SPLITS}...")

    X_train_raw = X.iloc[train_index]
    X_test_raw = X.iloc[test_index]

    y_train = y_binary[train_index]
    y_test = y_binary[test_index]

    # PREPROCESSING

    fold_preprocessor = copy.deepcopy(preprocessor)

    X_train_processed = fold_preprocessor.fit_transform(
        X_train_raw
    )

    X_test_processed = fold_preprocessor.transform(
        X_test_raw
    )

    if hasattr(X_train_processed, "toarray"):
        X_train_processed = X_train_processed.toarray()

    if hasattr(X_test_processed, "toarray"):
        X_test_processed = X_test_processed.toarray()

    # SELEKSI FITUR PEARSON

    selected_indices, correlations = pearson_feature_selection(
        X_train_processed,
        y_train,
        PEARSON_THRESHOLD
    )

    X_train_selected = X_train_processed[:, selected_indices]
    X_test_selected = X_test_processed[:, selected_indices]

    total_features = X_train_processed.shape[1]
    selected_features = len(selected_indices)

    reduction_percentage = (
        (total_features - selected_features)
        / total_features * 100
    )

    feature_records.append({
        "Fold": fold,
        "Jumlah Fitur Sebelum Seleksi": total_features,
        "Jumlah Fitur Sesudah Seleksi": selected_features,
        "Jumlah Fitur Dihapus": (
            total_features - selected_features
        ),
        "Persentase Pengurangan (%)": reduction_percentage
    })

    # DUA KONDISI EKSPERIMEN

    experiments = {
        "Tanpa Seleksi Pearson": (
            X_train_processed,
            X_test_processed
        ),

        "Dengan Seleksi Pearson": (
            X_train_selected,
            X_test_selected
        )
    }

    # PELATIHAN DAN EVALUASI MODEL

    for condition, (X_train_exp, X_test_exp) in experiments.items():

        scaler = StandardScaler()

        X_train_scaled = scaler.fit_transform(X_train_exp)
        X_test_scaled = scaler.transform(X_test_exp)

        for model_name, model in models.items():

            current_model = copy.deepcopy(model)

            current_model.fit(
                X_train_scaled,
                y_train
            )

            y_pred = current_model.predict(
                X_test_scaled
            )

            y_prob = current_model.predict_proba(
                X_test_scaled
            )[:, 1]

            # CONFUSION MATRIX

            cm = confusion_matrix(
                y_test,
                y_pred,
                labels=[0, 1]
            )

            tn, fp, fn, tp = cm.ravel()

            specificity = (
                tn / (tn + fp)
                if (tn + fp) > 0 else 0
            )

            # METRIK EVALUASI

            accuracy = accuracy_score(y_test, y_pred)

            precision = precision_score(
                y_test,
                y_pred,
                zero_division=0
            )

            recall = recall_score(
                y_test,
                y_pred,
                zero_division=0
            )

            balanced_acc = balanced_accuracy_score(
                y_test,
                y_pred
            )

            f1 = f1_score(
                y_test,
                y_pred,
                zero_division=0
            )

            auc = roc_auc_score(
                y_test,
                y_prob
            )

            results.append({
                "Fold": fold,
                "Kondisi": condition,
                "Model": model_name,
                "Accuracy": accuracy,
                "Precision": precision,
                "Recall": recall,
                "Specificity": specificity,
                "Balanced Accuracy": balanced_acc,
                "F1-Score": f1,
                "ROC-AUC": auc,
                "TP": tp,
                "TN": tn,
                "FP": fp,
                "FN": fn
            })

            # AGREGASI CONFUSION MATRIX

            key = (condition, model_name)

            if key not in confusion_totals:
                confusion_totals[key] = np.zeros(
                    (2, 2),
                    dtype=int
                )

            confusion_totals[key] += cm

            # DATA ROC

            fpr, tpr, _ = roc_curve(
                y_test,
                y_prob
            )

            roc_records.append({
                "Fold": fold,
                "Kondisi": condition,
                "Model": model_name,
                "FPR": fpr,
                "TPR": tpr,
                "AUC": auc
            })


# ============================================================
# 11. HASIL EVALUASI SETIAP FOLD
# ============================================================

df_results = pd.DataFrame(results)

print("\n" + "=" * 65)
print("HASIL EVALUASI SETIAP FOLD")
print("=" * 65)

display(df_results)


# ============================================================
# 12. RATA-RATA, STANDAR DEVIASI, DAN CONFIDENCE INTERVAL
# ============================================================

metric_columns = [
    "Accuracy",
    "Precision",
    "Recall",
    "Specificity",
    "Balanced Accuracy",
    "F1-Score",
    "ROC-AUC"
]

summary_records = []

for (condition, model_name), group in df_results.groupby(
    ["Kondisi", "Model"]
):

    for metric in metric_columns:

        values = group[metric].to_numpy()

        mean_value = np.mean(values)
        std_value = np.std(values, ddof=1)

        sem_value = stats.sem(values)

        ci = stats.t.interval(
            0.95,
            df=len(values) - 1,
            loc=mean_value,
            scale=sem_value
        )

        summary_records.append({
            "Kondisi": condition,
            "Model": model_name,
            "Metrik": metric,
            "Mean": mean_value,
            "Standar Deviasi": std_value,
            "CI 95% Bawah": ci[0],
            "CI 95% Atas": ci[1]
        })

df_summary = pd.DataFrame(summary_records)

print("\n" + "=" * 65)
print("RINGKASAN HASIL EVALUASI")
print("=" * 65)

display(df_summary)


# ============================================================
# 13. PERBANDINGAN JUMLAH FITUR
# ============================================================

df_features = pd.DataFrame(feature_records)

print("\nPerbandingan Jumlah Fitur Setiap Fold:")
display(df_features)

feature_summary = df_features[
    [
        "Jumlah Fitur Sebelum Seleksi",
        "Jumlah Fitur Sesudah Seleksi",
        "Jumlah Fitur Dihapus",
        "Persentase Pengurangan (%)"
    ]
].mean().to_frame(name="Rata-rata")

print("\nRata-rata Pengurangan Fitur:")
display(feature_summary)

# Grafik perbandingan jumlah fitur

plt.figure(figsize=(10, 6))

plt.plot(
    df_features["Fold"],
    df_features["Jumlah Fitur Sebelum Seleksi"],
    marker="o",
    label="Sebelum Seleksi"
)

plt.plot(
    df_features["Fold"],
    df_features["Jumlah Fitur Sesudah Seleksi"],
    marker="o",
    label="Sesudah Seleksi"
)

plt.title("Perbandingan Jumlah Fitur Sebelum dan Sesudah Seleksi")
plt.xlabel("Fold")
plt.ylabel("Jumlah Fitur")
plt.xticks(range(1, N_SPLITS + 1))
plt.legend()
plt.tight_layout()
plt.show()


# ============================================================
# 14. PERBANDINGAN KINERJA SEBELUM DAN SESUDAH SELEKSI
# ============================================================

comparison_table = df_summary.pivot_table(
    index=["Model", "Metrik"],
    columns="Kondisi",
    values="Mean"
).reset_index()

comparison_table["Selisih"] = (
    comparison_table["Dengan Seleksi Pearson"]
    - comparison_table["Tanpa Seleksi Pearson"]
)

print("\nPerbandingan Kinerja Model:")
display(comparison_table)

# Grafik perbandingan setiap metrik

for metric in metric_columns:

    plot_data = df_summary[
        df_summary["Metrik"] == metric
    ]

    plt.figure(figsize=(10, 6))

    sns.barplot(
        data=plot_data,
        x="Model",
        y="Mean",
        hue="Kondisi"
    )

    plt.title(f"Perbandingan {metric}")
    plt.xlabel("Algoritma")
    plt.ylabel(metric)
    plt.ylim(0, 1)
    plt.legend(title="Kondisi")
    plt.tight_layout()
    plt.show()


# ============================================================
# 15. ANALISIS CONFUSION MATRIX
# ============================================================

confusion_records = []

fig, axes = plt.subplots(
    1,
    3,
    figsize=(17, 5)
)

for ax, model_name in zip(axes, models.keys()):

    cm = confusion_totals[
        ("Dengan Seleksi Pearson", model_name)
    ]

    tn, fp, fn, tp = cm.ravel()

    sns.heatmap(
        cm,
        annot=True,
        fmt="d",
        cmap="Blues",
        cbar=False,
        ax=ax,
        xticklabels=["Prediksi Tidak", "Prediksi Ya"],
        yticklabels=["Aktual Tidak", "Aktual Ya"]
    )

    ax.set_title(model_name)
    ax.set_xlabel("Prediksi")
    ax.set_ylabel("Aktual")

plt.suptitle(
    "Confusion Matrix Setelah Seleksi Fitur Pearson"
)

plt.tight_layout()
plt.show()

# Tabel confusion matrix untuk kedua kondisi

for (condition, model_name), cm in confusion_totals.items():

    tn, fp, fn, tp = cm.ravel()

    confusion_records.append({
        "Kondisi": condition,
        "Model": model_name,
        "TN": tn,
        "FP": fp,
        "FN": fn,
        "TP": tp
    })

confusion_table = pd.DataFrame(confusion_records)

print("\nTabel Confusion Matrix:")
display(confusion_table)


# ============================================================
# 16. ANALISIS ROC-AUC
# ============================================================

mean_fpr = np.linspace(0, 1, 100)

roc_summary_records = []

fig, ax = plt.subplots(figsize=(9, 7))

for model_name in models.keys():

    selected_roc = [
        item for item in roc_records
        if item["Model"] == model_name
        and item["Kondisi"] == "Dengan Seleksi Pearson"
    ]

    tprs = []
    auc_values = []

    for item in selected_roc:

        interp_tpr = np.interp(
            mean_fpr,
            item["FPR"],
            item["TPR"]
        )

        interp_tpr[0] = 0.0

        tprs.append(interp_tpr)
        auc_values.append(item["AUC"])

    mean_tpr = np.mean(tprs, axis=0)
    std_tpr = np.std(tprs, axis=0)

    mean_auc = np.mean(auc_values)
    std_auc = np.std(auc_values, ddof=1)

    roc_summary_records.append({
        "Model": model_name,
        "Mean ROC-AUC": mean_auc,
        "Standar Deviasi ROC-AUC": std_auc
    })

    ax.plot(
        mean_fpr,
        mean_tpr,
        label=f"{model_name} (AUC = {mean_auc:.4f})"
    )

    lower = np.maximum(
        mean_tpr - std_tpr,
        0
    )

    upper = np.minimum(
        mean_tpr + std_tpr,
        1
    )

    ax.fill_between(
        mean_fpr,
        lower,
        upper,
        alpha=0.15
    )

ax.plot(
    [0, 1],
    [0, 1],
    linestyle="--",
    label="Random Classifier"
)

ax.set_title(
    "Mean ROC Curve Setelah Seleksi Fitur Pearson"
)

ax.set_xlabel("False Positive Rate")
ax.set_ylabel("True Positive Rate")
ax.legend(loc="lower right")
ax.grid(True)

plt.tight_layout()
plt.show()

roc_summary_df = pd.DataFrame(
    roc_summary_records
)

print("\nRingkasan ROC-AUC:")
display(roc_summary_df)


# ============================================================
# 17. ANALISIS STABILITAS MODEL
# ============================================================

stability_data = df_results[
    df_results["Kondisi"] == "Dengan Seleksi Pearson"
]

# Grafik stabilitas Accuracy

plt.figure(figsize=(10, 6))

sns.boxplot(
    data=stability_data,
    x="Model",
    y="Accuracy"
)

plt.title(
    "Stabilitas Accuracy Model dengan 10-Fold Cross Validation"
)

plt.xlabel("Algoritma")
plt.ylabel("Accuracy")
plt.tight_layout()
plt.show()

# Grafik stabilitas seluruh metrik

for metric in metric_columns:

    plt.figure(figsize=(10, 6))

    sns.boxplot(
        data=stability_data,
        x="Model",
        y=metric
    )

    plt.title(
        f"Stabilitas {metric} pada 10-Fold Cross Validation"
    )

    plt.xlabel("Algoritma")
    plt.ylabel(metric)
    plt.tight_layout()
    plt.show()


# ============================================================
# 18. ANALISIS KORELASI PEARSON
# ============================================================

# Korelasi ditampilkan dari data training Fold pertama.
# Seleksi fitur utama tetap dilakukan pada setiap fold.

first_train_index, first_test_index = next(
    skf.split(X, y_binary)
)

X_train_first = X.iloc[first_train_index]
y_train_first = y_binary[first_train_index]

preprocessor_corr = copy.deepcopy(preprocessor)

X_corr = preprocessor_corr.fit_transform(
    X_train_first
)

if hasattr(X_corr, "toarray"):
    X_corr = X_corr.toarray()

feature_names = preprocessor_corr.get_feature_names_out()

corr_values = []

for i in range(X_corr.shape[1]):

    if np.std(X_corr[:, i]) == 0:
        corr = 0
    else:
        corr = np.corrcoef(
            X_corr[:, i],
            y_train_first
        )[0, 1]

    if not np.isfinite(corr):
        corr = 0

    corr_values.append(corr)

correlation_table = pd.DataFrame({
    "Fitur": feature_names,
    "Korelasi Pearson": corr_values,
    "Korelasi Absolut": np.abs(corr_values)
})

correlation_table = correlation_table.sort_values(
    by="Korelasi Absolut",
    ascending=False
)

print("\nKorelasi Pearson Fitur terhadap Target:")
display(correlation_table)

# Grafik 20 fitur dengan korelasi absolut tertinggi

top_corr = correlation_table.head(20).sort_values(
    by="Korelasi Absolut"
)

plt.figure(figsize=(10, 8))

sns.barplot(
    data=top_corr,
    x="Korelasi Absolut",
    y="Fitur"
)

plt.title("20 Fitur dengan Korelasi Pearson Tertinggi")
plt.xlabel("Nilai Absolut Korelasi Pearson")
plt.ylabel("Fitur")
plt.tight_layout()
plt.show()


# ============================================================
# 18.1 VISUALISASI HEATMAP KORELASI PEARSON
# ============================================================

# Membuat DataFrame hasil preprocessing
df_corr = pd.DataFrame(
    X_corr,
    columns=feature_names,
    index=X_train_first.index
)

# Menambahkan kolom target
df_corr["y"] = y_train_first

# Menghitung matriks korelasi Pearson
pearson_matrix = df_corr.corr(method="pearson")

# Visualisasi heatmap
plt.figure(figsize=(18, 14))

sns.heatmap(
    pearson_matrix,
    cmap="coolwarm",
    vmin=-1,
    vmax=1,
    center=0,
    annot=False,
    linewidths=0.3,
    linecolor="white",
    cbar_kws={
        "label": "Koefisien Korelasi Pearson"
    }
)

plt.title(
    "Heatmap Korelasi Pearson Antarfitur Bank Marketing",
    fontsize=15,
    pad=15
)

plt.xlabel("Fitur")
plt.ylabel("Fitur")

plt.xticks(rotation=90, fontsize=8)
plt.yticks(rotation=0, fontsize=8)

plt.tight_layout()
plt.show()

# Tabel korelasi setiap fitur terhadap target

print("\nKorelasi Pearson Setiap Fitur terhadap Target:")

target_correlation = pearson_matrix["y"].drop("y")

target_correlation = target_correlation.sort_values(
    key=abs,
    ascending=False
)

display(
    target_correlation.to_frame(
        name="Korelasi Pearson terhadap Target"
    )
)


# ============================================================
# 19. RINGKASAN AKHIR HASIL ANALISIS
# ============================================================

print("\n" + "=" * 65)
print("ANALISIS KLASIFIKASI BANK MARKETING SELESAI")
print("=" * 65)

print("\n1. INFORMASI DATASET")
display(dataset_info)

print("\n2. DISTRIBUSI KELAS")
display(class_distribution)

print("\n3. JENIS FITUR")
display(feature_types)

print("\n4. STATISTIK DESKRIPTIF")
display(descriptive_stats)

print("\n5. PERBANDINGAN JUMLAH FITUR")
display(df_features)

print("\n6. RATA-RATA PENGURANGAN FITUR")
display(feature_summary)

print("\n7. RINGKASAN EVALUASI MODEL")
display(df_summary)

print("\n8. PERBANDINGAN KINERJA SEBELUM DAN SESUDAH SELEKSI")
display(comparison_table)

print("\n9. CONFUSION MATRIX")
display(confusion_table)

print("\n10. HASIL ROC-AUC")
display(roc_summary_df)

print("\n11. KORELASI PEARSON")
display(correlation_table)

print("\n12. MATRIKS KORELASI PEARSON")
display(pearson_matrix)

print("\nSeluruh proses analisis dan visualisasi telah selesai.")
```
